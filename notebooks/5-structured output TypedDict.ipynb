```python
import os
from dotenv import load_dotenv
load_dotenv()  # reads OLLAMA_MODEL from .env (set per machine)
from langchain_ollama import ChatOllama
from langchain.agents import create_agent
from langchain.tools import tool
llm = ChatOllama(model=os.getenv("OLLAMA_MODEL", "qwen3:8b"), temperature=0, reasoning=True)
@tool
def get_weather(city: str) -> str:
    """Get the current weather forecast for a given city."""
    return f"The weather in {city} is sunny and 24°C."

agent = create_agent(
    model=llm,
    tools=[get_weather],
    system_prompt="You are a helpful assistant. Use tools when needed.",
)
```

```python
resp = agent.invoke({"messages":[{"role":"user", "content":"Why do dogs bark?"}]})
print(resp)
```

```python
from typing_extensions import TypedDict
from langchain.agents import create_agent
class EmpInfo(TypedDict):
    """Employee info."""
    name: str
    email: str
    phone_no:str

agent = create_agent(model=llm, response_format=EmpInfo)
resp = agent.invoke({"messages":[{"role":"user", "content":"Extract contact info from: John Doe, john@example.com, (555) 123-4567"}]})

resp["structured_response"]
```

```python
from dataclasses import dataclass

@dataclass
class EmpInfoData:
    """Employee info."""
    name: str
    email: str
    phone_no:str

agent = create_agent(model=llm, response_format=EmpInfoData)
resp = agent.invoke({"messages":[{"role":"user", "content":"Extract contact info from: John Doe, john@example.com, (555) 123-4567"}]})

resp["structured_response"]
```
